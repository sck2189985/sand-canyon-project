```python
import pandas as pd
import numpy as np
```

```python
#cleaned facility data
rural_clinics = pd.read_excel("Agg Data FINAL/Panhandle_Rural_Health_Clinics.xlsx")
fqhc = pd.read_excel("Agg Data FINAL/Panhandle_FQHCs.xlsx")

rural_facilities = pd.DataFrame({
    "Facility_Name": rural_clinics["LEGALNAME"],
    "County": rural_clinics["CNTYNAME"],
    "City": rural_clinics["FAC_CITY"],
    "Address": rural_clinics["MDSDBA_FACILITY_ADDRESS"],
    "Longitude": rural_clinics["X"],
    "Latitude": rural_clinics["Y"],
    "Facility_Type": "Rural Health Clinic"})

fqhc_facilities = pd.DataFrame({
    "Facility_Name": fqhc["Site Name"],
    "County": fqhc["County Equivalent Name"],
    "City": fqhc["Site City"],
    "Address": fqhc["Site Address"],
    "Longitude": fqhc["Geocoding Artifact Address Primary X Coordinate"],
    "Latitude": fqhc["Geocoding Artifact Address Primary Y Coordinate"],
    "Facility_Type": "FQHC"})


all_facilities = pd.concat([rural_facilities, fqhc_facilities], ignore_index=True)
display(all_facilities)
```

```python
all_facilities["County"] = all_facilities["County"].str.upper()


print(all_facilities["County"].value_counts().sort_index())
print("Missing longitude:", all_facilities["Longitude"].isna().sum())
print("Missing latitude:", all_facilities["Latitude"].isna().sum())
```

```python
print(all_facilities[["Facility_Name", "County", "City"]].to_string(index=False))
```

```python
# add Sand Canyon to the facility list
sand_canyon = pd.DataFrame({
    "Facility_Name": ["Sand Canyon Health Center"],
    "County": ["BOX BUTTE"],
    "City": ["HEMINGFORD"],
    "Address": ["800 Box Butte Ave, Hemingford, NE 69348"],
    "Longitude": [-103.07309104622846],
    "Latitude": [42.32187009111592],
    "Facility_Type": ["Sand Canyon"]})

# current conditions include Sand Canyon
facilities_current = pd.concat(
    [all_facilities, sand_canyon],
    ignore_index=True)

# scenario without Sand Canyon
facilities_without = all_facilities.copy()

print("Current facilities:", len(facilities_current))
print("Facilities without Sand Canyon:", len(facilities_without))
```

```python
# adding sand canyon to the list of all facilities for simulation
all_facilities.to_excel("Agg Data FINAL/Panhandle_All_Facilities.xlsx", index=False)
```
