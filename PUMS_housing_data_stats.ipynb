```python
import pandas as pd
```

```python
pums_housing_df_filtered = pd.read_csv("PUMS Data FINAL/PUMS_housing_cleaned_data_FINAL.csv")
print(pums_housing_df_filtered.head())
print(pums_housing_df_filtered.shape)
print(pums_housing_df_filtered.columns.tolist())
print(pums_housing_df_filtered.dtypes)

#basic summary statistics
display(pums_housing_df_filtered.describe())
```

```python
print("Housing weight summary:")
print(pums_housing_df_filtered["WGTP"].describe())

print("Number of zero weights:")
print((pums_housing_df_filtered["WGTP"] == 0).sum())

print("Missing values by column:")
print(pums_housing_df_filtered.isna().sum().sort_values(ascending=False).head(50))

print("Percentage missing by column:")
print((pums_housing_df_filtered.isna().mean() * 100).sort_values(ascending=False).head(50))
```

```python
#of the total weight, what percentage is missing for each variable?

total_weight = pums_housing_df_filtered["WGTP"].sum()
weighted_missing = {}

for col in pums_housing_df_filtered.columns:
    missing_weight = pums_housing_df_filtered.loc[pums_housing_df_filtered[col].isna(), "WGTP"].sum()
    weighted_missing[col] = (missing_weight / total_weight) * 100

weighted_missing_df = pd.DataFrame({"Variable": weighted_missing.keys(), "Weighted Missing Percent": weighted_missing.values()})
weighted_missing_df = weighted_missing_df.sort_values("Weighted Missing Percent", ascending=False)

display(weighted_missing_df.head(50))
```
