```python
#load ACS raw excel file into pandas dataframe [read in ALL ACS raw data files] 

#dont forget to add all of the individual ACS raw data files to the "ACS Raw Data" folder in the same directory as this script 
#for this test ive only added a few of the ACS raw data files, but you should add all of them to clean all data



from pathlib import Path
import pandas as pd

data_folder = Path("ACS Raw Data")

#save the names of the files in a dictionary, where the keys are the names of the files and the values are the filenames
files = {
    "DPO2": "ACSDP5Y2024.DP02-2026-11-19T061427.csv",
    "DP03": "ACSDP5Y2024.DP03-2026-11-19T061350.csv",
    "DP04": "ACSDP5Y2024.DP04-2026-11-19T060311.csv",
    "DP05": "ACSDP5Y2024.DP05-2026-11-19T061508.csv",
    "B01001": "ACSDT5Y2024.B01001-2026-11-19T061855.csv",
    "B17001": "ACSDT5Y2024.B17001-2026-11-19T061640.csv",
    "B18101": "ACSDT5Y2024.B18101-2026-11-19T061720.csv",
}

#create a dictionary of dataframes, where the keys are the names of the files and the values are the dataframes
dfs = {
    name: pd.read_csv(data_folder / filename)
    for name, filename in files.items()
}

#now keep only the columns that contain "Estimate" in their name for each dataframe
estimate_dfs = {
    name: df.loc[:, df.columns.str.contains("Estimate", case=False)]
    for name, df in dfs.items()
}
```

```python
#unpivot tbales

id_col = "Label (Grouping)"

estimate_dfs = {
    name: df.loc[
        :,
        (df.columns == id_col)
        | df.columns.str.contains("Estimate", case=False)
    ]
    for name, df in dfs.items()
}

long_df = pd.concat(
    [
        df.melt(
            id_vars=["Label (Grouping)"],
            var_name="geography",
            value_name="value",
        ).assign(dataset=name)
        for name, df in estimate_dfs.items()
    ],
    ignore_index=True,
)

#now display each table so far
for name, table in estimate_dfs.items():
    print(name)
    display(table)
```

```python
#now extract the county and tract from the geography column

geography = long_df["geography"].astype("string")

long_df["tract"] = geography.str.extract(
    r"Census Tract\s+([0-9]+)",
    expand=False,
)

long_df["county"] = geography.str.extract(
    r";\s*([^;]+?)\s+County;",
    expand=False,
)

#now display each table so far
for name, table in estimate_dfs.items():
    print(name)
    display(table)
```

```python
#now create a dictionary with county fips codes (#double check on ACS website that these are correct one more time, i wrote them down from the ACS website but want to make sure they are correct)
county_fips = {
    "Banner": "007",
    "Box Butte": "013",
    "Cheyenne": "033",
    "Dawes": "045",
    "Deuel": "049",
    "Garden": "069",
    "Kimball": "105",
    "Morrill": "123",
    "Scotts Bluff": "157",
    "Sheridan": "161",
    "Sioux": "165",
}

long_df["county_fips"] = long_df["county"].map(county_fips)
```

```python
#now build GEOID column with brute force
#this is important because this will make the ACS data compatible with the CDC data
#rember that Nebraska state fips code is 31, so the GEOID will be 31 + county fips
# #+ tract code (tract code is the tract number with the decimal removed and padded with leading zeros to make it six digits long)
geography = long_df["geography"].astype("string")

long_df["tract"] = geography.str.extract(
    r"Census Tract\s+([0-9]+(?:\.[0-9]+)?)",
    expand=False,
)

long_df["county"] = geography.str.extract(
    r";\s*([^;]+?)\s+County;",
    expand=False,
)

long_df["county_fips"] = long_df["county"].map(county_fips).astype("string")

#census tract codes are six digits with the decimal removed (e.g. 1.01 -> 000101).
long_df["tract_code"] = (
    long_df["tract"]
    .str.replace(".", "", regex=False)
    .str.zfill(6)
)

valid = long_df["county_fips"].notna() & long_df["tract_code"].notna()
long_df["GEOID"] = (
    "31" + long_df["county_fips"] + long_df["tract_code"]
).where(valid)

columns_to_show = [
    "geography", "county", "county_fips",
    "tract", "tract_code", "GEOID", "value",
]

for name in estimate_dfs:
    print(f"{name} — first 10 rows")
    display(
        long_df.loc[long_df["dataset"] == name, columns_to_show].head(10)
    )
```

```python
folder = Path("ACS Raw Data")

poverty_file = list(folder.glob("ACSST5Y2024.S1701*.csv"))[0]
insurance_file = list(folder.glob("ACSST5Y2024.S2701*.csv"))[0]

poverty = pd.read_csv(poverty_file)
insurance = pd.read_csv(insurance_file)
```

```python
# getting overall poverty percentages
poverty_values = poverty.iloc[0]
poverty_values = poverty_values[poverty_values.index.str.endswith("!!Percent below poverty level!!Estimate")]

# get the overall uninsured percentages
insurance_values = insurance.iloc[0]
insurance_values = insurance_values[insurance_values.index.str.endswith("!!Percent Uninsured!!Estimate")]


poverty_data = pd.DataFrame({"Tract": poverty_values.index.str.replace("!!Percent below poverty level!!Estimate", "", regex=False),"Poverty_Rate": poverty_values.values})

insurance_data = pd.DataFrame({"Tract": insurance_values.index.str.replace("!!Percent Uninsured!!Estimate", "", regex=False),"Uninsured_Rate": insurance_values.values})

# combining the two variables
acs_need = poverty_data.merge(insurance_data, on="Tract")

# removing percent signs and converting to numbers
acs_need["Poverty_Rate"] = pd.to_numeric(acs_need["Poverty_Rate"].str.replace("%", ""), errors="coerce")
acs_need["Uninsured_Rate"] = pd.to_numeric(acs_need["Uninsured_Rate"].str.replace("%", ""), errors="coerce")
```

```python
age_file = list(folder.glob("ACSST5Y2024.S0101*.csv"))[0]
age = pd.read_csv(age_file)

# finding the row for people age 65 and over
age65 = age[age["Label (Grouping)"].str.strip() == "65 years and over"]
```

```python
# overall age 65 and over percentages
age_values = age65.iloc[0]
age_values = age_values[age_values.index.str.endswith("!!Percent!!Estimate")]

age_data = pd.DataFrame({"Tract": age_values.index.str.replace("!!Percent!!Estimate", "", regex=False),"Age_65_Plus": age_values.values})

age_data["Age_65_Plus"] = pd.to_numeric(age_data["Age_65_Plus"].str.replace("%", ""),errors="coerce")

# adding age 65+ to the existing dataset
acs_need = acs_need.merge(age_data, on="Tract")
display(acs_need)
```

```python
disability_file = list(folder.glob("ACSST5Y2024.S1810*.csv"))[0]
disability = pd.read_csv(disability_file)
```

```python
# overall disability percentages
disability_values = disability.iloc[0]
disability_values = disability_values[disability_values.index.str.endswith("!!Percent with a disability!!Estimate")]

disability_data = pd.DataFrame({"Tract": disability_values.index.str.replace("!!Percent with a disability!!Estimate", "", regex=False), "Disability_Rate": disability_values.values})

disability_data["Disability_Rate"] = pd.to_numeric(disability_data["Disability_Rate"].str.replace("%", ""),errors="coerce")

# adding disability 
acs_need = acs_need.merge(disability_data, on="Tract")
```

```python
# loading the DP03 table for SNAP data
snap_file = list(folder.glob("ACSDP5Y2024.DP03*.csv"))[0]
DP03 = pd.read_csv(snap_file)
```

```python
# find the SNAP row
snap = DP03[DP03["Label (Grouping)"].str.contains("With Food Stamp/SNAP benefits", case=False, na=False)]
```

```python
# getting SNAP percentage for each tract
snap_row = snap.iloc[0]

snap_values = snap_row[snap_row.index.str.endswith("!!Percent")]
snap_data = pd.DataFrame({"Tract": snap_values.index.str.replace("!!Percent", "", regex=False),"SNAP_Rate": snap_values.values})

# cleaning the percentage values
snap_data["SNAP_Rate"] = (snap_data["SNAP_Rate"].astype(str).str.replace("%", "", regex=False))
snap_data["SNAP_Rate"] = pd.to_numeric(snap_data["SNAP_Rate"], errors="coerce")

acs_need = acs_need.merge(snap_data, on="Tract", how="left")

print(acs_need.head())
print(acs_need["SNAP_Rate"].isna().sum())
```

```python
acs_need = pd.read_csv("Agg Data FINAL/Panhandle_ACS_Community_Need.csv")
display(acs_need.head())
```

```python
# standardizing the community need variables, using z-scores to make them comparable across different scales and distributions. This allows for a more accurate assessment of community needs by normalizing the data.
need_variables = ["Poverty_Rate", "Uninsured_Rate", "Age_65_Plus", "Disability_Rate", "SNAP_Rate"]

for variable in need_variables:
    acs_need[variable + "_Z"] = (
        acs_need[variable] - acs_need[variable].mean()
    ) / acs_need[variable].std()
```

```python
display(acs_need.head())
```

```python
standardized_variables = ["Poverty_Rate_Z", "Uninsured_Rate_Z", "Age_65_Plus_Z", "Disability_Rate_Z", "SNAP_Rate_Z"]

# calculating the overall community need index
acs_need["Community_Need_Index"] = (acs_need[standardized_variables].mean(axis=1))

# sorting tracts from highest to lowest community need
acs_need = acs_need.sort_values("Community_Need_Index", ascending=False)
display(acs_need[["Tract", "Community_Need_Index"]])
```
