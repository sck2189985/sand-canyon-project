```python
import pandas as pd

pums_housing_df = pd.read_csv("PUMS raw data files\PUMS_housing_rawdata.csv")
print(pums_housing_df.head())       # Preview rows
print(pums_housing_df.shape)        # Rows and columns
print(pums_housing_df.columns)      # Column names

#save the cleaned data to a new CSV file
pums_housing_df.to_csv("PUMS cleaning data files\PUMS_housing_cleaned_data.csv", index=False)


####now do the same for the PUMS person data
pums_person_df = pd.read_csv("PUMS raw data files\PUMS_person_rawdata.csv")
import pandas as pd

print(pums_person_df.head())       # Preview rows
print(pums_person_df.shape)        # Rows and columns
print(pums_person_df.columns)      # Column names

#save the cleaned data to a new CSV file
pums_person_df.to_csv("PUMS cleaning data files\PUMS_person_cleaned_data.csv", index=False)
```

```python
#create a new dataframe with summary statistics
pums_housing_df_stats=pums_housing_df.describe()  # Summary statistics

#display the summary statistics
display(pums_housing_df_stats)

#check the data type of the summary statistics dataframe
print(type(pums_housing_df_stats))

#save the summary statistics to a new CSV file
pums_housing_df_stats.to_csv("PUMS cleaning data files\PUMS_housing_cleaned_data_stats.csv", index=False)



####now do the same for the PUMS person data
#create a new dataframe with summary statistics
pums_person_df_stats=pums_person_df.describe()  # Summary statistics

#display the summary statistics
display(pums_person_df_stats)

#check the data type of the summary statistics dataframe
print(type(pums_person_df_stats))

#save the summary statistics to a new CSV file
pums_person_df_stats.to_csv("PUMS cleaning data files\PUMS_person_cleaned_data_stats.csv", index=False)
```

```python
#now need to filter out only the rows that have a value of 0100 in the column "PUMA"

#might have to fix this later if other descripter codes need the 0 but it looks like the PUMA column transfered over as an integer and removed
#the leading 0. So we can just filter for 100 instead of 0100. Double check later that other identifiers that start with 0 are not affected by this.

pums_housing_df_filtered = pums_housing_df[pums_housing_df['PUMA'] == 100]


#create a new dataframe with summary statistics for the filtered data
pums_housing_df_filtered_stats=pums_housing_df_filtered.describe()  # Summary statistics    

#save both the filtered data and the summary statistics to new CSV files
pums_housing_df_filtered.to_csv("PUMS cleaning data files\PUMS_housing_cleaned_data_filtered.csv", index=False)

pums_housing_df_filtered_stats.to_csv("PUMS cleaning data files\PUMS_housing_cleaned_data_filtered_stats.csv", index=False)

display(pums_housing_df_filtered)
display(pums_housing_df_filtered_stats)
```
