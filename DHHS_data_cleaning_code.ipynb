```python
import pandas as pd
import numpy as np
```

```python
rural_clinics = pd.read_csv("PUMS raw data files/DHHS_Rural_Health_Clinics.csv")

# counties included in our project
panhandle_counties = ["BANNER", "BOX BUTTE", "CHEYENNE", "DAWES", "DEUEL", "GARDEN", "KIMBALL", "MORRILL", "SCOTTS BLUFF", "SHERIDAN", "SIOUX"]
panhandle_clinics = rural_clinics[rural_clinics["CNTYNAME"].isin(panhandle_counties)].copy()

panhandle_clinics = panhandle_clinics[["LEGALNAME", "CNTYNAME", "FAC_CITY", "MDSDBA_FACILITY_ADDRESS", "X", "Y", "GIS_Status"]]
display(panhandle_clinics)
```

```python
#checking the number of clinics and missing coordinates
print("Number of clinics:", len(panhandle_clinics))
print("Missing longitude:", panhandle_clinics["X"].isna().sum())
print("Missing latitude:", panhandle_clinics["Y"].isna().sum())

#saving the cleaned clinic data as an excel file as the other data is!
panhandle_clinics.to_excel("Agg Data FINAL/Panhandle_Rural_Health_Clinics.xlsx", index=False)
```
