```python
#now we will check the using the Successive Difference Replication (SDR) Formula for Calculating Uncertainty [basically 
# well be using the replicate weights to calculate the standard error of the weighted mean age]

import numpy as np
import pandas as pd

age_col = "AGEP"      
weight_col = "PWGTP"   
rep_cols = [f"PWGTP{i}" for i in range(1, 81)]


cols = [age_col, weight_col, *rep_cols]
data = pums_person_df_filtered[cols].apply(pd.to_numeric, errors="coerce").dropna()
data = data[data[weight_col] > 0]

estimate = np.average(data[age_col], weights=data[weight_col])

replicate_estimates = [
    np.average(data[age_col], weights=data[col])
    for col in rep_cols
]

variance = (4 / 80) * sum(
    (rep_estimate - estimate) ** 2
    for rep_estimate in replicate_estimates
)
standard_error = np.sqrt(variance)

print("Weighted mean age:", estimate)
print("Standard error:", standard_error)
```

```python
#this will show distrbution for all of the variables that were selctted[basically the variables that we are interested in for our analysis] and 
# it will also show the percentage of missing values for each variable. For the continuous variables it will also show the weighted mean and standard deviation.
df = pums_person_df_filtered
weight_col = "PWGTP"

variables = [
    "DIS", "HINS4", "HINS3", "HINS2", "HINS1",
    "DREM", "DDRS", "DEAR", "DEYE", "DPHY",
    "JWMNP", "JWRIP", "JWTRNS", "HICOV",
    "POVPIP", "AGEP", "PINCP",
]

continuous_vars = {"JWMNP", "POVPIP","PINCP", "AGEP"}

if weight_col not in df.columns:
    raise KeyError(f"{weight_col} is missing from the DataFrame")

missing_cols = [col for col in variables if col not in df.columns]
if missing_cols:
    print("These columns are missing and will be skipped:", missing_cols)

missingness_rows = []

for col in variables:
    if col not in df.columns:
        continue

    data = df[[col, weight_col]].copy()
    data[col] = pd.to_numeric(data[col], errors="coerce")
    data[weight_col] = pd.to_numeric(data[weight_col], errors="coerce")
    data = data.dropna(subset=[weight_col])
    data = data[data[weight_col] > 0]

    total_weight = data[weight_col].sum()
    missing_weight = data.loc[data[col].isna(), weight_col].sum()
    nonmissing_weight = data.loc[data[col].notna(), weight_col].sum()

    missingness_rows.append({
        "variable": col,
        "estimated_missing": missing_weight,
        "missing_percent": 100 * missing_weight / total_weight if total_weight else np.nan,
        "estimated_nonmissing": nonmissing_weight,
        "nonmissing_percent": 100 * nonmissing_weight / total_weight if total_weight else np.nan,
    })

    print(f"\n{col}")
    print(f"Weighted missing:     {missing_weight:,.0f} ({100 * missing_weight / total_weight:.2f}%)")
    print(f"Weighted non-missing: {nonmissing_weight:,.0f} ({100 * nonmissing_weight / total_weight:.2f}%)")

    valid = data.dropna(subset=[col])

    if valid.empty:
        print("No non-missing values to summarize.")
        continue

    if col in continuous_vars:
        mean = np.average(valid[col], weights=valid[weight_col])
        std = np.sqrt(np.average((valid[col] - mean) ** 2, weights=valid[weight_col]))
        print(f"Non-missing weighted mean: {mean:.2f}")
        print(f"Non-missing weighted standard deviation: {std:.2f}")
    else:
        distribution = (
            valid.groupby(col)[weight_col]
            .sum()
            .rename("estimated_people")
            .to_frame()
        )
        distribution["percent_of_nonmissing"] = (
            100 * distribution["estimated_people"] / nonmissing_weight
        )
        display(distribution)
        distribution.to_csv(
            f"PUMS stats files/PUMS_person_{col}_distribution.csv"
        )

PUMS_person_missingness_summary = pd.DataFrame(missingness_rows).set_index("variable")
display(PUMS_person_missingness_summary)
PUMS_person_missingness_summary.to_csv("PUMS stats files/PUMS_person_missingness_summary.csv")
```

```python
#now do the Successive Difference Replication (SDR) Formula for Calculating Uncertainty for all of the variables that were selctted[basically 
# well be using the replicate weights to calculate the standard error of the weighted mean age] <-for more info on how this calucalted see PUMS documentation

import numpy as np
import pandas as pd
from IPython.display import display
from pathlib import Path

df = pums_person_df_filtered

variables = [
    "DIS", "HINS4", "HINS3", "HINS2", "HINS1",
    "DREM", "DDRS", "DEAR", "DEYE", "DPHY",
    "JWMNP", "JWRIP", "JWTRNS", "HICOV", "PINCP",
     "POVPIP", "AGEP",
]

main_weight = "PWGTP"
replicate_weights = [f"PWGTP{i}" for i in range(1, 81)]

#treat these as measurements where a weighted mean is useful.
continuous_vars = {"JWMNP", "POVPIP","PINCP", "AGEP"}

required_weights = [main_weight, *replicate_weights]
missing_weights = [c for c in required_weights if c not in df.columns]
if missing_weights:
    raise KeyError(f"Missing person-weight columns: {missing_weights}")

missing_variables = [c for c in variables if c not in df.columns]
if missing_variables:
    print("Variables missing from the DataFrame; skipping:", missing_variables)

for variable in variables:
    if variable not in df.columns:
        continue

    data = df[[variable, *required_weights]].copy()
    data = data.apply(pd.to_numeric, errors="coerce")
    data = data.dropna()
    data = data[data[main_weight] > 0]

    if data.empty:
        print(f"\n{variable}: no usable rows; skipping.")
        continue

    print(f"\n{variable}")

    if variable in continuous_vars:
        estimate = np.average(data[variable], weights=data[main_weight])
        replicate_estimates = [
            np.average(data[variable], weights=data[weight])
            for weight in replicate_weights
        ]

        variance = (4 / len(replicate_weights)) * sum(
            (rep_estimate - estimate) ** 2
            for rep_estimate in replicate_estimates
        )

        table = pd.DataFrame({
            "weighted_mean": [estimate],
            "standard_error": [np.sqrt(variance)],
        }, index=[variable])

    else:
        categories = data[variable].drop_duplicates().tolist()
        rows = []

        for category in categories:
            in_category = data[variable] == category
            estimate = data.loc[in_category, main_weight].sum()

            replicate_estimates = [
                data.loc[in_category, weight].sum()
                for weight in replicate_weights
            ]
            variance = (4 / len(replicate_weights)) * sum(
                (rep_estimate - estimate) ** 2
                for rep_estimate in replicate_estimates
            )

            percent = estimate / data[main_weight].sum() * 100
            replicate_percents = [
                data.loc[in_category, weight].sum()
                / data[weight].sum() * 100
                for weight in replicate_weights
            ]
            percent_variance = (4 / len(replicate_weights)) * sum(
                (rep_percent - percent) ** 2
                for rep_percent in replicate_percents
            )

            rows.append({
                variable: category,
                "estimated_people": estimate,
                "count_standard_error": np.sqrt(variance),
                "percent": percent,
                "percent_standard_error": np.sqrt(percent_variance),
            })

        PUMS_SDR_person_table = pd.DataFrame(rows).set_index(variable)

    display(PUMS_SDR_person_table)
    output_dir = Path("PUMS stats files\PUMS_SDR_stats")
    output_dir.mkdir(parents=True, exist_ok=True)

    save_table = table if variable in continuous_vars else PUMS_SDR_person_table
    save_table.to_csv(output_dir / f"PUMS_person_{variable}_SDR.csv", index=True)
```
