
### FQHC Data Cleaning and Findings

The HRSA Health Center Service Delivery and Look-Alike Sites dataset was filtered to include active sites located in the 11 Nebraska Panhandle counties.

After filtering, 3 records remain:

- 1 site in Dawes County (Chadron).
- 2 sites in Scotts Bluff County (Gering).

The cleaned dataset includes facility names, addresses, county locations, operating hours, and geographic coordinates. These coordinates will be useful when calculating the estimated travel distances and times to healthcare facilities.

(As of initial analysis) Note that:

- The two Gering records have very similar coordinates but different addresses and facility names. These should probably be reviewed to determine whether they represent separate locations?
- The CAPWN Dental Clinic in Chadron is specifically identified as a dental clinic, so it should not automatically be treated as a general primary-care facility...may need to be omitted based on what we decide.
- One record is missing operating hours...it was kept because the missing value does not affect its location information but good to know!

These results provide additional information about healthcare facility locations in the panhandle! The cleaned data will be combined with the rural health clinic data to help establish the existing healthcare facility locations for the access simulation.


```python
import pandas as pd
import numpy as np
```

```python
#loading the UDS csv file
UDS_data = pd.read_csv('PUMS raw data files/Health_Center_Service_Delivery_and_LookAlike_Sites.csv')

# counties included in our project
panhandle_counties = ["BANNER", "BOX BUTTE", "CHEYENNE", "DAWES", "DEUEL", "GARDEN", "KIMBALL", "MORRILL", "SCOTTS BLUFF", "SHERIDAN", "SIOUX"]

# keeping only active federally qualified health centers in Nebraska
panhandle_fqhc = UDS_data[
    (UDS_data["Site State Abbreviation"] == "NE") &
    (UDS_data["Site Status Description"] == "Active") &
    (UDS_data["County Equivalent Name"].str.upper().isin(panhandle_counties))].copy()

# keeping the info we need for the simulation
panhandle_fqhc = panhandle_fqhc[["Site Name", "Site Address", "Site City", "County Equivalent Name", "Operating Hours per Week", "Geocoding Artifact Address Primary X Coordinate", "Geocoding Artifact Address Primary Y Coordinate"]]
display(panhandle_fqhc)
```
